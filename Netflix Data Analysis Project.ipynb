**Load the data from csv file**

```python
netflix_data = pd.read_csv("netflix data.csv")
```

# **Visualize Data**

**Plot a Bar Plot to compare Top 5 Countries with Most Movies**

```python
top_5_countries_by_number_of_movies_released = netflix_data[
    netflix_data['type'] == 'Movie'
].groupby('country').agg(
        Number_Of_Releases = ('show_id', 'count')
).sort_values(by='Number_Of_Releases', ascending=False).reset_index().head()
```

```python
top_5_countries_by_number_of_movies_released
```

```python
sns.set_style('whitegrid')
plt.figure(figsize=(12, 3))
plt.title("Top 5 Countries by Number of Movies Released")
plt.xlabel("Country")
plt.ylabel("Number of Released")
sns.barplot(
    data=top_5_countries_by_number_of_movies_released,
    x='country',
    y='Number_Of_Releases',
    hue='country',
    width=.25
)
plt.show()
```

**Plot a Bar Plot to compare Top 5 Countries with Most TV Shows**

```python
top_5_countries_by_number_of_tv_shows_released = netflix_data[
    netflix_data['type'] == 'TV Show'
].groupby('country').agg(
    Number_Of_Releases = ('show_id', 'count')
).sort_values(by='Number_Of_Releases', ascending=False).head().reset_index()
```

```python
top_5_countries_by_number_of_tv_shows_released
```

```python
plt.figure(figsize=(12, 3))
plt.xlabel("Country")
plt.ylabel("Number of Released")
plt.title("Top 5 Countries By Number of TV Shows Released")
sns.barplot(
    data=top_5_countries_by_number_of_tv_shows_released,
    x='country',
    y='Number_Of_Releases',
    hue='country',
    width=.25
)
plt.show()
```

**Plot a Pie Chart to show the Percentages of Movies Released by Top 10 Genres**

```python
top_10_genres_by_number_of_movies_released = netflix_data[
    netflix_data['type'] == 'Movie'
].groupby('listed_in').agg(
    Number_Of_Released = ('show_id', 'count')
).sort_values(by='Number_Of_Released', ascending=False).head(10).reset_index()
```

```python
top_10_genres_by_number_of_movies_released
```

```python
plt.title("Top 10 Genres By Number of Movies Released")
plt.pie(
    top_10_genres_by_number_of_movies_released['Number_Of_Released'],
    labels=top_10_genres_by_number_of_movies_released['listed_in'],
    autopct="%.2f%%"
)
plt.show()
```

**Plot a Pie Chart to show the Percentages of TV Shows Released by Top 10 Genres**

```python
top_10_genres_by_number_of_tv_shows_released = netflix_data[netflix_data['type'] == 'TV Show'].groupby(
    'listed_in').agg(
        Number_Of_Released = ('show_id', 'count')
).sort_values(by='Number_Of_Released', ascending=False).head(10).reset_index()
```

```python
top_10_genres_by_number_of_tv_shows_released
```

```python
plt.title("Top 10 Genres By Number of TV Shows Released")
plt.pie(
    top_10_genres_by_number_of_tv_shows_released['Number_Of_Released'],
    labels=top_10_genres_by_number_of_tv_shows_released['listed_in'],
    autopct="%.2f%%"
)
plt.show()
```

**Plot a Line Plot to compare the Trend Line of Numbers of Movies and TV Shows released across all the years**

```python
number_of_released_by_release_year = netflix_data.groupby(['release_year', 'type']).agg(
    Number_Of_Released = ('show_id', 'count')
).reset_index()
```

```python
number_of_released_by_release_year
```

```python
plt.figure(figsize=(12, 3))
plt.title("Number of Released By Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Released")
sns.lineplot(
    data=number_of_released_by_release_year,
    x='release_year',
    y='Number_Of_Released',
    hue='type'
)
plt.show()
```

**Plot a Bar Plot to compare number of Movies and TV Shows released by last 10 years**

```python
number_of_released_by_last_10_years = netflix_data[
    netflix_data['release_year'] >= max_release_year - 10
].groupby(['release_year', 'type']).agg(
    Number_Of_Released = ('show_id', 'count')
).reset_index()
```

```python
number_of_released_by_last_10_years
```

```python
plt.figure(figsize=(12, 3))
plt.title("Number of Released By Last 10 Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Released")
sns.barplot(
    data=number_of_released_by_last_10_years,
    x='release_year',
    y='Number_Of_Released',
    hue='type'
)
plt.show()
```
